```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class AgentState(TypedDict):
    name: str
    age: int
    skills: list[str]
    result: str
```

```python
def evaluate_name(state: AgentState)-> AgentState:
    """Taking name of the user to result"""

    state['result'] = f"Your name is {state['name']}"
    return state

def evaluate_age(state: AgentState)-> AgentState:
    """Taking age of the user to result"""
    state['result'] = f"{state['result']} Your age is {state['age']}"
    return state

def evaluate_skills(state: AgentState)-> AgentState:
    """Taking skills of the user to result"""
    skills = ""
    for skill in state['skills']:
        skills = f'{skills}{skill}, '
    state['result'] = f"{state['result']} and you are skilled in {skills}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node('evaluate_name', evaluate_name)
graph.add_node('evaluate_age', evaluate_age)
graph.add_node('evaluate_skills', evaluate_skills)

graph.set_entry_point('evaluate_name')
graph.add_edge('evaluate_name', 'evaluate_age')
graph.add_edge('evaluate_age', 'evaluate_skills')
graph.add_edge('evaluate_skills', END)

workflow = graph.compile()
```

```python
workflow.invoke({
    'name': 'Visal Singh',
    'age': 21,
    'skills': ['Js', 'Ts', 'Py']
})
```
