```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class PlayerStats(TypedDict):
    runs: int
    balls: int
    fours: int
    sixes: int
    strike_rate: float
    per_boundry_balls: float
    runs_by_boundry: float
```

```python
def calc_strike_rate(state: PlayerStats):
    return {
        "strike_rate": state["runs"] / state["balls"]
    }


def calc_boundry_per_ball(state: PlayerStats):
    boundries = state["fours"] + state["sixes"]

    return {
        "per_boundry_balls": state["balls"] / boundries
    }


def calc_runs_by_boundry(state: PlayerStats):
    return {
        "runs_by_boundry": (
            state["fours"] * 4
            + state["sixes"] * 6
        )
    }

def assosiate_stats(state: PlayerStats) -> PlayerStats:
    return state
```

```python
graph = StateGraph(PlayerStats)

graph.add_node("calc_strike_rate", calc_strike_rate)
graph.add_node("calc_boundry_per_ball", calc_boundry_per_ball)
graph.add_node("calc_runs_by_boundry", calc_runs_by_boundry)
graph.add_node("assosiate", assosiate_stats)


graph.add_edge(START, "calc_strike_rate")
graph.add_edge(START, "calc_boundry_per_ball")
graph.add_edge(START, "calc_runs_by_boundry")

graph.add_edge("calc_strike_rate", "assosiate")
graph.add_edge("calc_boundry_per_ball", "assosiate")
graph.add_edge("calc_runs_by_boundry", "assosiate")

graph.add_edge("assosiate", END)


workflow = graph.compile()
```

```python
result = workflow.invoke({
    'runs': 200,
    'balls': 130,
    'fours': 20,
    'sixes': 6
})

print(result)
```
